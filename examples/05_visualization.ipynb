# Visualization with LAGO

This notebook shows how to create visualizations of temporal communities using `LongitudinalModulesPlot`.

The visualization shows:
- **Nodes** on the y-axis
- **Time** on the x-axis
- **Communities** as colored rectangles

**Note:** Requires matplotlib to be installed.

```python
import matplotlib.pyplot as plt
from lago import LinkStream, lago_modules
from lago.viz import LongitudinalModulesPlot

# Enable inline plotting
%matplotlib inline
```

## Create Sample Data

Let's create a linkstream with evolving communities.

```python
def create_sample_data():
    """Create sample linkstream and communities."""
    ls = LinkStream()
    ls.add_links([
        # Group 1: nodes 0, 1, 2
        (0, 1, 0), (1, 2, 0), (0, 2, 0),
        (0, 1, 1), (1, 2, 1),
        (0, 1, 2), (1, 2, 2),
        # Group 2: nodes 3, 4
        (3, 4, 0), (3, 4, 1), (3, 4, 2),
        # Bridge at time 3
        (2, 3, 3),
        # Evolution: groups merge
        (0, 3, 4), (1, 4, 4),
        (0, 1, 4), (3, 4, 4),
    ])
    communities = lago_modules(ls)
    return ls, communities

ls, communities = create_sample_data()
print(f"Detected {communities.nb_modules} communities")
```

## 1. Basic Plot

Create a basic visualization (simplest approach).

```python
# Create plot from TimeModules (linkstream is also required)
plot = LongitudinalModulesPlot(communities, linkstream=ls)

# Draw (will display inline)
plot.draw()
plt.show()
```

## 2. Plot with Edge Activity

Show when edges are active between nodes.

```python
# Create plot with linkstream (required for edge visualization)
plot = LongitudinalModulesPlot(communities, linkstream=ls)

# Configure edges to show activity
plot.configure_edges(
    show_edges=False,       # Don't show arc connections
    show_activity=True,     # Show when edges are active
    activity_alpha=0.5,
)

plot.draw()
plt.show()
```

## 3. Customized Plot

Create a fully customized visualization with labels and styling.

```python
# Create plot with custom size
plot = LongitudinalModulesPlot(communities, linkstream=ls, width=1200, height=600)

# Configure nodes
plot.configure_nodes(
    labels={0: "Alice", 1: "Bob", 2: "Carol", 3: "Dave", 4: "Eve"},
    auto_ordering=True,  # Reorder nodes to minimize crossings
    fontsize=10,
)

# Configure edges
plot.configure_edges(
    show_activity=True,
    activity_width=0.8,
    activity_height=0.6,
    activity_alpha=0.3,
)

# Configure communities
plot.configure_modules(
    color_palette="tab10",
    height=0.8,
)

# Configure display
plot.configure_display(
    padding_bottom=0.5,
    padding_top=0.5,
    show_xlabel=True,
    show_ylabel=True,
)

plot.draw()
plt.show()
```

## 4. Focused View

Focus on specific nodes or communities.

```python
plot = LongitudinalModulesPlot(communities, linkstream=ls)

# Focus on specific nodes
plot.configure_nodes(
    focus=[0, 1, 2],        # Highlight nodes 0, 1, 2
    linewidth=0.3,          # Thin lines for non-focused
    linewidth_focus=1.0,    # Thick lines for focused
)

plot.draw()
plt.show()
```

## 5. Plot from Manual Partition

Visualize manually defined communities.

```python
from lago import TimeModules

ls_simple = LinkStream()
ls_simple.add_links([
    (0, 1, 0), (1, 2, 0),
    (0, 1, 1), (2, 3, 1),
])

# Define communities as dict
manual_communities = TimeModules({
    "Group A": {(0, 0), (1, 0), (0, 1), (1, 1)},
    "Group B": {(2, 0), (2, 1), (3, 1)},
})

plot = LongitudinalModulesPlot(manual_communities, linkstream=ls_simple)
plot.draw()
plt.show()
```

## 6. Advanced Matplotlib Customization

Get the matplotlib axis for further customization.

```python
plot = LongitudinalModulesPlot(communities, linkstream=ls)

# Get the matplotlib figure and axis
fig, ax = plot.draw(return_ax=True)

# Add custom annotations
ax.annotate(
    "Communities merge here",
    xy=(3.5, 2),
    xytext=(4, 3),
    arrowprops=dict(arrowstyle="->", color="red"),
    fontsize=9,
    color="red",
)

# Add title
ax.set_title("Temporal Community Evolution", fontsize=14)

plt.tight_layout()
plt.show()
```
