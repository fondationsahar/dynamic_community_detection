# Advanced Visualization API

This notebook demonstrates the full visualization API for `LongitudinalModulesPlot`.

For simpler examples, see [05_visualization.ipynb](05_visualization.ipynb) first.

The API consolidates configuration into 4 main methods:
1. `configure_nodes()` - All node-related settings
2. `configure_edges()` - All edge-related settings
3. `configure_modules()` - All community-related settings
4. `configure_display()` - Global display settings (axis, labels, padding)

````python
import matplotlib.pyplot as plt
from lago import LinkStream, lago_modules
from lago.viz import LongitudinalModulesPlot

%matplotlib inline
````

## 1. Visualizing lago_modules Results

Most common use case: detect communities and visualize them.

This is the recommended approach for most users.

````python
# Create linkstream
ls = LinkStream()
ls.add_links([
    # Dense group 1
    (0, 1, 0), (1, 2, 0), (0, 2, 0),
    (0, 1, 1), (1, 2, 1),
    # Dense group 2
    (3, 4, 0), (4, 5, 0),
    (3, 4, 1), (4, 5, 1),
    # Evolution
    (2, 3, 2),
    (0, 1, 2), (3, 4, 2),
])

# Detect communities
communities = lago_modules(ls)

print(f"Detected {communities.nb_modules} communities")
````

````python
# Create plot with TimeModules (simplest API)
plot = LongitudinalModulesPlot(communities, linkstream=ls, width=1200, height=600)

# Configure with simplified API
plot.configure_nodes(
    labels={0: "Alice", 1: "Bob", 2: "Carol", 3: "Dave", 4: "Eve", 5: "Frank"},
    auto_ordering=True,
    fontsize=10,
)

plot.configure_edges(
    show_activity=True,
    activity_alpha=0.4,
)

plot.configure_modules(
    color_palette="tab10",
    height=0.75,
)

plot.configure_display(
    padding_bottom=0.5,
    padding_top=0.5,
    show_xlabel=True,
    show_ylabel=True,
)

plot.draw()
plt.show()
````

## 2. Full Simplified API Example

The simplified API follows a clear structure:
1. Create plot with TimeModules (and optional linkstream)
2. `configure_nodes()` - labels, focus, ordering, style
3. `configure_edges()` - visibility, activity markers, style
4. `configure_modules()` - colors, focus, style
5. `configure_display()` - axis, padding, labels

````python
# Create sample data
ls_simple = LinkStream()
ls_simple.add_links([
    (0, 1, 0), (1, 2, 0),
    (0, 1, 1), (2, 3, 1),
    (0, 1, 2), (2, 3, 2),
])
communities_simple = lago_modules(ls_simple)
````

````python
# Create plot with custom size
plot = LongitudinalModulesPlot(communities_simple, linkstream=ls_simple, width=1600, height=800)

# === Configure nodes ===
plot.configure_nodes(
    nodes=[0, 1, 2, 3],              # Explicit node list (optional)
    labels={0: "A", 1: "B", 2: "C", 3: "D"},  # Node labels
    focus=None,                       # Optional: list of node IDs to highlight
    auto_ordering=True,              # Reorder nodes to minimize edge crossings
    fontsize=8,
    label_padding=-15,
    linewidth=0.25,
    linewidth_focus=0.5,
)

# === Configure edges ===
plot.configure_edges(
    show_edges=False,        # Don't show arc edges
    show_activity=True,      # Show edge activity markers
    activity_width=0.9,
    activity_height=0.5,
    activity_alpha=0.3,
)

# === Configure modules ===
plot.configure_modules(
    max_shown=10,            # Maximum communities to display
    color_palette="tab20",   # Color palette
    height=0.75,             # Community rectangle height
)

# === Configure display ===
plot.configure_display(
    padding_bottom=0.5,
    padding_top=1.0,
    show_xlabel=False,
    show_ylabel=False,
    show_yticks=True,
)

# === Draw ===
plot.draw()
plt.show()
````

## 3. Configuration Options Reference

Here's a comprehensive reference of all configuration options.

### configure_nodes() Options

```python
plot.configure_nodes(
    nodes=None,              # List of node IDs to display (None = all)
    labels=None,             # Dict mapping node_id -> label string
    focus=None,              # List of node IDs to highlight
    auto_ordering=False,     # Auto-reorder to minimize crossings
    fontsize=8,              # Font size for labels
    label_padding=-15,       # Padding between label and node line
    linewidth=0.25,          # Width of node timeline
    linewidth_focus=0.5,     # Width of focused node timeline
)
```

### configure_edges() Options

```python
plot.configure_edges(
    show_edges=False,        # Show arc connections between nodes
    show_activity=True,      # Show edge activity markers
    activity_width=0.9,      # Width of activity marker
    activity_height=0.5,     # Height of activity marker
    activity_alpha=0.3,      # Transparency of activity marker
)
```

### configure_modules() Options

```python
plot.configure_modules(
    max_shown=10,                    # Max communities to show
    color_palette="tab10",           # Matplotlib color palette
    height=0.75,                     # Rectangle height (0-1)
    focus_modules=None,              # Dict: {module_id: color}
    show_unfocused=True,             # Show non-focused modules
    unfocused_style='lighter',       # 'lighter' or 'greyed'
)
```

**Available color palettes:** `tab10`, `tab20`, `Set1`, `Set2`, `Set3`, `Pastel1`, `Pastel2`, `Dark2`, `Accent`

### configure_display() Options

```python
plot.configure_display(
    padding_bottom=0.5,      # Space below plot
    padding_top=0.5,         # Space above plot
    show_xlabel=True,        # Show time axis label
    show_ylabel=True,        # Show node axis label
    show_yticks=True,        # Show node tick marks
)
```

## 4. Experiment with Different Settings

Try modifying these parameters to see their effects!

````python
# Create a plot and experiment with settings
plot = LongitudinalModulesPlot(communities, linkstream=ls, width=1400, height=700)

# Try different color palettes
plot.configure_modules(color_palette="Set2", height=0.9)

# Try focusing on specific nodes
plot.configure_nodes(focus=[0, 1], linewidth_focus=1.5)

# Try different edge activity styles
plot.configure_edges(show_activity=True, activity_alpha=0.6, activity_height=0.7)

plot.draw()
plt.show()
````

## API Comparison Summary

**Original API (scattered):**
- `set_nodes()` + `auto_node_ordering()` + `set_node_style()` + `set_focus_nodes()`
- `toggle_edges()` + `toggle_edge_activity()` + `set_edge_style()` + `set_edge_activity_style()`
- `set_communities()` + `set_color_palette()` + `set_community_style()` + ...
- `set_axis_padding()` + `set_labels()`

**Simplified API (consolidated):**
- `configure_nodes()` - All node settings in one call
- `configure_edges()` - All edge settings in one call
- `configure_modules()` - All community settings in one call
- `configure_display()` - All display settings in one call

Both APIs are fully compatible - you can mix them as needed.
